# Demo: Sensory Data Analysis

This notebook demonstrates a simple workflow for sensory data visualization.

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate synthetic sensory data
np.random.seed(42)
n_samples = 100

# Physical measurements
viscosity = np.random.uniform(0.1, 2.0, n_samples)
friction = 0.3 * viscosity + np.random.normal(0, 0.1, n_samples)

# Simulated sensory scores
thickness_score = 2 + 3 * viscosity + np.random.normal(0, 0.5, n_samples)
smoothness_score = 8 - 5 * friction + np.random.normal(0, 0.3, n_samples)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Plot 1: Viscosity vs Thickness
axes[0].scatter(viscosity, thickness_score, alpha=0.6, c='steelblue')
axes[0].set_xlabel('Viscosity (Pa·s)')
axes[0].set_ylabel('Thickness Score')
axes[0].set_title('Viscosity → Thickness Perception')

# Plot 2: Friction vs Smoothness
axes[1].scatter(friction, smoothness_score, alpha=0.6, c='coral')
axes[1].set_xlabel('Friction Coefficient')
axes[1].set_ylabel('Smoothness Score')
axes[1].set_title('Friction → Smoothness Perception')

plt.tight_layout()
plt.show()
```

## Correlation Analysis

```python
r_thickness = np.corrcoef(viscosity, thickness_score)[0, 1]
r_smoothness = np.corrcoef(friction, smoothness_score)[0, 1]

print(f"Correlation (Viscosity - Thickness): r = {r_thickness:.3f}")
print(f"Correlation (Friction - Smoothness): r = {r_smoothness:.3f}")
```
